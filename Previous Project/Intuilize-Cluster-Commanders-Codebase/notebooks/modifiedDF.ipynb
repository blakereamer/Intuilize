```python
import pandas as pd
```

```python
# Read original data frame

path_to_raw_data = '../data/raw/Intuilize_MNSU_ACME_SalesData.csv'
original_df = pd.read_csv(path_to_raw_data)
```

```python
######### Data Engineering ########

# Select required columns
selected_columns = ['CustomerID','CustomerName','ExtPrice','ExtCost','SalesQty','UnitPrice','UnitCost','UnitSize']
df = original_df[selected_columns]

# Add new columns including Profit, ProfitMargin, PriceCost_Ratio
df['Profit'] = df['ExtPrice'] - df['ExtCost']
df['ProfitMargin'] = df['Profit'] / df['ExtPrice'].replace(0, 1)
df['ProfitCost_Ratio'] = df['UnitPrice'] / df['UnitCost'].replace(0, 1)

# Save to the new CSV file (newDF)
df.to_csv("../data/raw/newDF.csv")
```
