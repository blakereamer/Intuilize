```python
from feature_creation_monthly import process_file
import os
import pandas as pd
import numpy as np
```

```python
path_to_data = '../data/raw/Intuilize_MNSU_ACME_SalesData.csv'
```

```python
data = pd.read_csv(path_to_data)
```

```python
data['SalesDate'] = pd.to_datetime(data['SalesDate'])
data["Month"] = data["SalesDate"].dt.month
data["Year"] = data["SalesDate"].dt.year
# Calculate profit
data["Profit"] = data["ExtPrice"] - data["ExtCost"]
```

```python
data = data.drop(data[data["Year"].isin([2021, 2025])].index)
```

```python
unique_customers = data['CustomerName'].unique()
name_to_alias = {name: f"Customer {i+1}" for i, name in enumerate(unique_customers)}
data['customer_alias'] = data['CustomerName'].map(name_to_alias)
```

```python
_2022_and_2023_data = data[data["Year"].isin([2022, 2023])]
_2024_data = data[data["Year"] == 2024]
#split 2022 & 2023 data into Monthly chunks
_2022_data = _2022_and_2023_data[_2022_and_2023_data["Year"] == 2022]
_2023_data = _2022_and_2023_data[_2022_and_2023_data["Year"] == 2023]

_2022_1_data = _2022_data[_2022_data["Month"].isin([1,2,3,4,5,6])]

_2022_2_data = _2022_data[_2022_data["Month"].isin([7,8,9,10,11,12])]

_2023_1_data = _2023_data[_2023_data["Month"].isin([1,2,3,4,5,6])]

_2023_2_data = _2023_data[_2023_data["Month"].isin([7,8,9,10,11,12])]
```

```python
customer_metrics = data.groupby(["CustomerID", "CustomerName"]).agg(
    TotalRevenue=("ExtPrice", "sum"),
    TotalProfit=("Profit", "sum"),
    PurchaseFrequency=("OrderNumber", "nunique"),
).reset_index()
```

```python
if not os.path.exists('../data/cleaned'):
    os.makedirs('../data/cleaned')

if not os.path.exists('../data/cleaned/halfyeardata'):
    os.makedirs('../data/cleaned/halfyeardata')

if not os.path.exists('../data/cleaned/halfyearmetrics'):
    os.makedirs('../data/cleaned/halfyearmetrics')

_2022_and_2023_data.to_csv("../data/cleaned/train_data.csv")
_2024_data.to_csv("../data/cleaned/test_data.csv")
customer_metrics.to_csv("../data/cleaned/customer_metrics.csv")
data.to_csv("../data/cleaned/raw_data_extra_features.csv") # removed 2021 and 2025 and have profit, month, and year

_2022_1_data.to_csv("../data/cleaned/halfyeardata/2022_1_data.csv")
_2022_2_data.to_csv("../data/cleaned/halfyeardata/2022_2_data.csv")

_2023_1_data.to_csv("../data/cleaned/halfyeardata/2023_1_data.csv")
_2023_2_data.to_csv("../data/cleaned/halfyeardata/2023_2_data.csv")
```

```python
import os
import pandas as pd

def interleave_halfyear_files(input_dir, output_file):
    # Define the filenames (adjust these if your filenames differ)
    file_2022_first = os.path.join(input_dir, "customer_metrics_2022_1_data.csv")
    file_2022_second = os.path.join(input_dir, "customer_metrics_2022_2_data.csv")
    file_2023_first = os.path.join(input_dir, "customer_metrics_2023_1_data.csv")
    file_2023_second = os.path.join(input_dir, "customer_metrics_2023_2_data.csv")
    
    # Read each file into a DataFrame and add Year and Half_Year columns
    df_2022_first = pd.read_csv(file_2022_first)
    df_2022_first['Year'] = 2022
    df_2022_first['Half_Year'] = 1

    df_2022_second = pd.read_csv(file_2022_second)
    df_2022_second['Year'] = 2022
    df_2022_second['Half_Year'] = 2

    df_2023_first = pd.read_csv(file_2023_first)
    df_2023_first['Year'] = 2023
    df_2023_first['Half_Year'] = 1

    df_2023_second = pd.read_csv(file_2023_second)
    df_2023_second['Year'] = 2023
    df_2023_second['Half_Year'] = 2
    
    # Determine maximum number of rows across the four DataFrames
    n_max = max(len(df_2022_first), len(df_2022_second), len(df_2023_first), len(df_2023_second))
    
    # Create a list to store the interleaved rows
    records = []
    for i in range(n_max):
        if i < len(df_2022_first):
            records.append(df_2022_first.iloc[i])
        if i < len(df_2022_second):
            records.append(df_2022_second.iloc[i])
        if i < len(df_2023_first):
            records.append(df_2023_first.iloc[i])
        if i < len(df_2023_second):
            records.append(df_2023_second.iloc[i])
    
    # Create the combined DataFrame
    combined_df = pd.DataFrame(records)
    combined_df.to_csv(output_file, index=False)
    print(f"Combined file written to {output_file}")

# Example usage:
input_dir = "../data/cleaned/halfyearmetrics"
output_file = "../data/cleaned/halfyearmetrics/combined_halfyear_metrics.csv"
interleave_halfyear_files(input_dir, output_file)
```
